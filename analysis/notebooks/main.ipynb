# PDX combination therapy: model review
This notebook reads the observed cohort and the reproducible pipeline outputs. Run `make data` and `make fit` from the repository root first. The fit implementation lives in `src/pdxmodel/`; no results are recalculated by hand here.

```python
import json
from pathlib import Path
import pandas as pd

root = Path.cwd().resolve()
analysis = root if (root / 'data/processed/pdx_combo.csv').exists() else root / 'analysis'
data = pd.read_csv(analysis / 'data/processed/pdx_combo.csv')
results = json.loads((analysis / 'outputs/results.json').read_text(encoding='utf-8'))
display(data.groupby(['pdx_id', 'treatment']).size().unstack())
display(results['summary'])
```

## Interpretation
The full fit estimates a positive interaction, but the Bliss-null model is not rejected. Both the trajectory-block bootstrap and refitted profile likelihood cover zero. Structured residuals and carrying capacities at bounds make the sign of interaction uncertain. See `MODEL.md` for equations and assumptions, and `outputs/figures/` for diagnostics.
